```python
# Read the CSV with default quoting enabled to keep columns aligned, while skipping bad lines (like an unfinished EOF)
df = pd.read_csv("/content/fraudTest.csv", on_bad_lines='skip')
df.head()
```

```python
print("Shape:", df.shape)
print("Missing values:", df.isna().sum().sum())
print(df[["amt", "city_pop"]].describe())
```

```python
counts = df["is_fraud"].value_counts()
print(counts)
print((counts / len(df) * 100).round(2))
counts.plot(kind="bar", title="0 = Legitimate, 1 = Fraud")
plt.show()
```

```python
import numpy as np
import pandas as pd

# Define variables if they were deleted
NUM = ["amt", "city_pop", "hour", "age", "distance_km"]
CAT = ["category", "gender"]
FEATURES = NUM + CAT

def engineer(df_input):
    d = pd.DataFrame(index=df_input.index)
    t = pd.to_datetime(df_input["trans_date_trans_time"], errors='coerce')
    dob = pd.to_datetime(df_input["dob"], errors='coerce')
    d["amt"] = pd.to_numeric(df_input["amt"], errors='coerce')
    d["city_pop"] = pd.to_numeric(df_input["city_pop"], errors='coerce')
    d["hour"] = t.dt.hour
    d["age"] = (t - dob).dt.days / 365.25

    # Haversine distance
    lat1, lon1 = map(np.radians, [pd.to_numeric(df_input["lat"], errors='coerce'), pd.to_numeric(df_input["long"], errors='coerce')])
    lat2, lon2 = map(np.radians, [pd.to_numeric(df_input["merch_lat"], errors='coerce'), pd.to_numeric(df_input["merch_long"], errors='coerce')])
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    d["distance_km"] = 6371 * 2 * np.arcsin(np.sqrt(a))

    d["category"] = df_input["category"]
    d["gender"] = df_input["gender"]
    return d[FEATURES]

# Keep only valid binary fraud targets and drop NaN/shifted rows
clean_df = df[df["is_fraud"].isin([0.0, 1.0, 0, 1])].dropna(subset=["is_fraud"]).copy()

X = engineer(clean_df)
y = clean_df["is_fraud"].astype(int)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
print("Fraud rate train:", y_train.mean().round(4), "| test:", y_test.mean().round(4))
```

```python
preprocess = ColumnTransformer([
    ("num", StandardScaler(), NUM),
    ("cat", OneHotEncoder(handle_unknown="ignore"), CAT),
])
model = Pipeline([
    ("preprocess", preprocess),
    ("clf", LogisticRegression(class_weight="balanced", max_iter=1000)),
])
model.fit(X_train, y_train)
```

```python
proba = model.predict_proba(X_test)[:, 1]
pred = (proba >= 0.5).astype(int)

ConfusionMatrixDisplay(confusion_matrix(y_test, pred),
                       display_labels=["Legit", "Fraud"]).plot()
plt.show()
print(classification_report(y_test, pred, digits=3))
print("ROC-AUC:", round(roc_auc_score(y_test, proba), 3))
```
